```python
import pandas as pd

team_stats = pd.read_csv("../../XGBoost-Model/Model_data/team_stats.csv")

team_stats.head()
team_stats.columns.tolist()
```

```python
team_stats_valid = (
    team_stats
    .groupby("gameid")
    .filter(lambda x: len(x) == 2)
)
team_stats_valid.groupby("gameid").size().value_counts()
```

```python
dataset_xgb = []
```

```python
for gameid, partido in team_stats_valid.groupby("gameid"):

    if len(partido) != 2:
        continue

    equipo1 = partido.iloc[0]
    equipo2 = partido.iloc[1]


    fila = {

        "winrate_diff":
            equipo1.winrate - equipo2.winrate,

        #"gamelength":
            #equipo1.gamelength - equipo2.gamelength,

        "gold15_diff":
            equipo1.gold15 - equipo2.gold15,

        "xp15_diff":
            equipo1.xp15 - equipo2.xp15,

        "cs15_diff":
            equipo1.cs15 - equipo2.cs15,

        #"first_blood_diff":
            #equipo1.firstblood - equipo2.firstblood,

        #"first_tower_diff":
            #equipo1.firsttower - equipo2.firsttower,

        #"first_dragon_diff":
            #equipo1.firstdragon - equipo2.firstdragon,

        #"first_herald_diff":
            #equipo1.firstherald - equipo2.firstherald,

        "dragon_diff":
            equipo1.dragons - equipo2.dragons,

        "baron_diff":
            equipo1.barons - equipo2.barons,

        "tower_diff":
            equipo1.towers - equipo2.towers,


        "target":
            equipo1.result

    }


    dataset_xgb.append(fila)
```

```python
xgb_df = pd.DataFrame(dataset_xgb)

xgb_df.to_csv(
    "../../XGBoost-Model/Model_data/worlds_xgb_dataset.csv",
    index=False
)

print(xgb_df.head())
team_stats_valid.result.value_counts(normalize=True)
xgb_df["target"].value_counts()
```
